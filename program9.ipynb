```python
# ============================================================
# EXPERIMENT 9
# IMAGE GENERATION USING DIFFUSION MODELS
# ============================================================

!pip install diffusers transformers accelerate torch pillow matplotlib -q

import torch
from diffusers import StableDiffusionPipeline
import matplotlib.pyplot as plt


# ============================================================
# 1. LOAD PRETRAINED STABLE DIFFUSION MODEL
# ============================================================

model_id = "runwayml/stable-diffusion-v1-5"

print("Loading Stable Diffusion model...")

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
    if torch.cuda.is_available()
    else torch.float32
)


# ============================================================
# 2. SELECT DEVICE
# ============================================================

device = "cuda" if torch.cuda.is_available() else "cpu"

pipe = pipe.to(device)

print("Device:", device)


# ============================================================
# 3. TEXT PROMPT
# ============================================================

prompt = """
A futuristic smart city with flying cars,
green buildings, robots assisting people,
highly detailed, realistic, 4K quality.
"""


# ============================================================
# 4. GENERATE IMAGE
# ============================================================

print("\nGenerating image...")

image = pipe(prompt).images[0]


# ============================================================
# 5. SAVE GENERATED IMAGE
# ============================================================

image.save("generated_image.png")


# ============================================================
# 6. DISPLAY GENERATED IMAGE
# ============================================================

plt.figure(figsize=(8, 8))

plt.imshow(image)

plt.axis("off")

plt.title("Generated Image")

plt.show()


# ============================================================
# 7. RESULT
# ============================================================

print("\nImage successfully generated and saved as generated_image.png")

print("\n")
print("=" * 60)
print("RESULT")
print("=" * 60)

print(
    "Thus, an image generation application was successfully "
    "implemented using a pretrained Stable Diffusion model. "
    "The application generated high-quality images from "
    "natural-language prompts and demonstrated the capability "
    "of diffusion models for AI-based image synthesis."
)
```
