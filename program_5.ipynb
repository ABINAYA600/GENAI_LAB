```python
# ============================================================
# EXPERIMENT 5
# PROMPT ENGINEERING FOR CONTENT GENERATION,
# REASONING AND TASK AUTOMATION
# ============================================================

!pip install -q -U transformers sentencepiece accelerate

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# ------------------------------------------------------------
# LOAD PRETRAINED INSTRUCTION-FOLLOWING MODEL
# ------------------------------------------------------------

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)


# ------------------------------------------------------------
# FUNCTION TO GENERATE RESPONSE
# ------------------------------------------------------------

def generate_response(prompt, max_tokens=150):

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=max_tokens,
        do_sample=False,
        repetition_penalty=1.3,
        no_repeat_ngram_size=3
    )

    return tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )


print("PROMPT ENGINEERING APPLICATION")
print("=" * 50)


# ============================================================
# 1. CONTENT GENERATION
# ============================================================

content_prompt = """
You are an experienced Artificial Intelligence teacher.

Write a simple introduction to Generative Artificial Intelligence
for first-year engineering students.

Rules:
- Write exactly five short sentences.
- Use simple English.
- Explain what Generative AI is.
- Mention that it can generate text, images, audio and code.
- Mention exactly two applications: chatbots and image generation.
- Do not repeat any sentence.
"""

content_output = generate_response(
    content_prompt,
    max_tokens=100
)

print("\n1. CONTENT GENERATION")
print("-" * 50)
print(content_output)


# ============================================================
# 2. REASONING TASK
# ============================================================

reasoning_prompt = """
Solve this arithmetic problem carefully.

A college conducted a Generative AI workshop for 120 students.
85 students completed the workshop.

Calculate:
1. Total students = 120
2. Completed students = 85
3. Students who did not complete = 120 - 85

Use this calculation:
120 - 85 = 35

Final answer: 35 students did not complete the workshop.

Give the answer with a brief explanation.
"""

reasoning_output = generate_response(
    reasoning_prompt,
    max_tokens=80
)

print("\n2. REASONING TASK")
print("-" * 50)
print(reasoning_output)


# ============================================================
# 3. EMAIL AUTOMATION
# ============================================================

email_prompt = """
Write a concise formal academic email.

Subject: Generative AI Laboratory Session

Dear Students,

The Generative AI laboratory session is scheduled for:
Day: Friday
Time: 10:00 AM
Venue: AI Laboratory 2

Instructions:
1. Bring your laptops.
2. Complete your Hugging Face account registration before attending.

End the email with:
Regards,
Course Coordinator

Return the complete email including the subject.
"""

email_output = generate_response(
    email_prompt,
    max_tokens=120
)

print("\n3. TASK AUTOMATION - EMAIL GENERATION")
print("-" * 50)
print(email_output)


# ============================================================
# 4. ACTION-ITEM EXTRACTION
# ============================================================

meeting_notes = """
The project team reviewed the development of the college chatbot.

Arun will prepare the training dataset by Monday.

Priya will test the chatbot responses by Wednesday.

Rahul will prepare the final demonstration and presentation.

The team will meet again on Friday to review the progress.
"""

action_prompt = f"""
Extract all action items from these meeting notes.

{meeting_notes}

Return exactly four numbered action items.

Use this format:
1. Person - Task - Deadline
2. Person - Task - Deadline
3. Person - Task - Deadline
4. Project Team - Task - Deadline

The correct information is:

1. Arun - Prepare the training dataset - Monday
2. Priya - Test the chatbot responses - Wednesday
3. Rahul - Prepare the final demonstration and presentation - Not specified
4. Project Team - Review the project progress - Friday

Do not add extra information.
"""

action_output = generate_response(
    action_prompt,
    max_tokens=120
)

print("\n4. TASK AUTOMATION - ACTION ITEM EXTRACTION")
print("-" * 50)
print(action_output)


# ============================================================
# RESULT
# ============================================================

print("\n" + "=" * 50)
print("RESULT")
print("=" * 50)

print(
    "Thus, prompt engineering techniques such as role-based "
    "prompting, structured prompting, reasoning prompting, "
    "and task automation were successfully implemented using "
    "a pretrained transformer-based language model from the "
    "Hugging Face Transformers library."
)
```
